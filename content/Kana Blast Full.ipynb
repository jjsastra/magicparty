```python
import ipywidgets as widgets
from IPython.display import display, clear_output, Javascript, HTML
import random

# ==================== FULL KANA DICTIONARIES ====================
HIRAGANA = {
    # Basic
    'あ': 'a', 'い': 'i', 'う': 'u', 'え': 'e', 'お': 'o',
    'か': 'ka', 'き': 'ki', 'く': 'ku', 'け': 'ke', 'こ': 'ko',
    'さ': 'sa', 'し': 'shi', 'す': 'su', 'せ': 'se', 'そ': 'so',
    'た': 'ta', 'ち': 'chi', 'つ': 'tsu', 'て': 'te', 'と': 'to',
    'な': 'na', 'に': 'ni', 'ぬ': 'nu', 'ね': 'ne', 'の': 'no',
    'は': 'ha', 'ひ': 'hi', 'ふ': 'fu', 'へ': 'he', 'ほ': 'ho',
    'ま': 'ma', 'み': 'mi', 'む': 'mu', 'め': 'me', 'も': 'mo',
    'や': 'ya', 'ゆ': 'yu', 'よ': 'yo',
    'ら': 'ra', 'り': 'ri', 'る': 'ru', 'れ': 're', 'ろ': 'ro',
    'わ': 'wa', 'を': 'wo', 'ん': 'n',

    # Dakuten
    'が': 'ga', 'ぎ': 'gi', 'ぐ': 'gu', 'げ': 'ge', 'ご': 'go',
    'ざ': 'za', 'じ': 'ji', 'ず': 'zu', 'ぜ': 'ze', 'ぞ': 'zo',
    'だ': 'da', 'ぢ': 'ji', 'づ': 'zu', 'で': 'de', 'ど': 'do',
    'ば': 'ba', 'び': 'bi', 'ぶ': 'bu', 'べ': 'be', 'ぼ': 'bo',

    # Handakuten
    'ぱ': 'pa', 'ぴ': 'pi', 'ぷ': 'pu', 'ぺ': 'pe', 'ぽ': 'po',

    # Small yōon
    'きゃ': 'kya', 'きゅ': 'kyu', 'きょ': 'kyo',
    'しゃ': 'sha', 'しゅ': 'shu', 'しょ': 'sho',
    'ちゃ': 'cha', 'ちゅ': 'chu', 'ちょ': 'cho',
    'にゃ': 'nya', 'にゅ': 'nyu', 'にょ': 'nyo',
    'ひゃ': 'hya', 'ひゅ': 'hyu', 'ひょ': 'hyo',
    'みゃ': 'mya', 'みゅ': 'myu', 'みょ': 'myo',
    'りゃ': 'rya', 'りゅ': 'ryu', 'りょ': 'ryo',
    'ぎゃ': 'gya', 'ぎゅ': 'gyu', 'ぎょ': 'gyo',
    'じゃ': 'ja',  'じゅ': 'ju',  'じょ': 'jo',
    'びゃ': 'bya', 'びゅ': 'byu', 'びょ': 'byo',
    'ぴゃ': 'pya', 'ぴゅ': 'pyu', 'ぴょ': 'pyo'
}

KATAKANA = {
    # Basic
    'ア': 'a', 'イ': 'i', 'ウ': 'u', 'エ': 'e', 'オ': 'o',
    'カ': 'ka', 'キ': 'ki', 'ク': 'ku', 'ケ': 'ke', 'コ': 'ko',
    'サ': 'sa', 'シ': 'shi', 'ス': 'su', 'セ': 'se', 'ソ': 'so',
    'タ': 'ta', 'チ': 'chi', 'ツ': 'tsu', 'テ': 'te', 'ト': 'to',
    'ナ': 'na', 'ニ': 'ni', 'ヌ': 'nu', 'ネ': 'ne', 'ノ': 'no',
    'ハ': 'ha', 'ヒ': 'hi', 'フ': 'fu', 'ヘ': 'he', 'ホ': 'ho',
    'マ': 'ma', 'ミ': 'mi', 'ム': 'mu', 'メ': 'me', 'モ': 'mo',
    'ヤ': 'ya', 'ユ': 'yu', 'ヨ': 'yo',
    'ラ': 'ra', 'リ': 'ri', 'ル': 'ru', 'レ': 're', 'ロ': 'ro',
    'ワ': 'wa', 'ヲ': 'wo', 'ン': 'n',

    # Dakuten
    'ガ': 'ga', 'ギ': 'gi', 'グ': 'gu', 'ゲ': 'ge', 'ゴ': 'go',
    'ザ': 'za', 'ジ': 'ji', 'ズ': 'zu', 'ゼ': 'ze', 'ゾ': 'zo',
    'ダ': 'da', 'ヂ': 'ji', 'ヅ': 'zu', 'デ': 'de', 'ド': 'do',
    'バ': 'ba', 'ビ': 'bi', 'ブ': 'bu', 'ベ': 'be', 'ボ': 'bo',

    # Handakuten
    'パ': 'pa', 'ピ': 'pi', 'プ': 'pu', 'ペ': 'pe', 'ポ': 'po',

    # Small yōon
    'キャ': 'kya', 'キュ': 'kyu', 'キョ': 'kyo',
    'シャ': 'sha', 'シュ': 'shu', 'ショ': 'sho',
    'チャ': 'cha', 'チュ': 'chu', 'チョ': 'cho',
    'ニャ': 'nya', 'ニュ': 'nyu', 'ニョ': 'nyo',
    'ヒャ': 'hya', 'ヒュ': 'hyu', 'ヒョ': 'hyo',
    'ミャ': 'mya', 'ミュ': 'myu', 'ミョ': 'myo',
    'リャ': 'rya', 'リュ': 'ryu', 'リョ': 'ryo',
    'ギャ': 'gya', 'ギュ': 'gyu', 'ギョ': 'gyo',
    'ジャ': 'ja',  'ジュ': 'ju',  'ジョ': 'jo',
    'ビャ': 'bya', 'ビュ': 'byu', 'ビョ': 'byo',
    'ピャ': 'pya', 'ピュ': 'pyu', 'ピョ': 'pyo'
}

ALL_ROMAJI = list(set(list(HIRAGANA.values()) + list(KATAKANA.values())))

# ==================== THE GAME ====================
class KanaBlast:
    def __init__(self):
        self.score = 0
        self.streak = 0
        self.current_kana = ""
        self.current_romaji = ""
        self.pool = list(HIRAGANA.items()) + list(KATAKANA.items())
        
        self.title = widgets.HTML("<h1 style='color:#ff00aa; text-align:center;'>🐼 KANA BLAST! FULL EDITION 🐼</h1>")
        
        self.mode_selector = widgets.Dropdown(
            options=['Hiragana Only', 'Katakana Only', 'Both (Mixed)'],
            value='Both (Mixed)', description='Mode:'
        )
        self.mode_selector.observe(self.change_mode, names='value')
        
        self.kana_display = widgets.HTML(value="", layout=widgets.Layout(height='160px'))
        
        self.status = widgets.HTML(value="<b style='font-size:26px;color:#00ff00;'>Score: 0 &nbsp;&nbsp;🔥 Streak: 0</b>")
        
        self.feedback = widgets.Output(layout=widgets.Layout(height='70px'))
        
        self.buttons = []
        self.hbox = widgets.HBox([], layout=widgets.Layout(justify_content='center', flex_flow='row wrap'))
        for _ in range(4):
            btn = widgets.Button(description="", layout=widgets.Layout(width='130px', height='65px', margin='6px'), button_style='info')
            btn.on_click(self.check_answer)
            self.buttons.append(btn)
            self.hbox.children += (btn,)
        
        self.speak_btn = widgets.Button(description="🔊 Speak it!", button_style='success', layout=widgets.Layout(width='220px', height='50px'))
        self.speak_btn.on_click(self.speak)
        
        self.next_btn = widgets.Button(description="Next Card ➡️", button_style='primary', layout=widgets.Layout(width='220px', height='50px'))
        self.next_btn.on_click(self.new_card)
        
        self.game = widgets.VBox([
            self.title,
            self.mode_selector,
            self.kana_display,
            self.status,
            self.feedback,
            self.hbox,
            widgets.HBox([self.speak_btn, self.next_btn], layout=widgets.Layout(justify_content='center', margin='15px'))
        ], layout=widgets.Layout(align_items='center'))
        
        display(self.game)
        self.new_card(None)
    
    def change_mode(self, change):
        if change['new'] == 'Hiragana Only':
            self.pool = list(HIRAGANA.items())
        elif change['new'] == 'Katakana Only':
            self.pool = list(KATAKANA.items())
        else:
            self.pool = list(HIRAGANA.items()) + list(KATAKANA.items())
        self.new_card(None)
    
    def new_card(self, b):
        self.current_kana, self.current_romaji = random.choice(self.pool)
        
        wrongs = random.sample([r for r in ALL_ROMAJI if r != self.current_romaji], 3)
        choices = [self.current_romaji] + wrongs
        random.shuffle(choices)
        
        self.kana_display.value = f"<div style='font-size:140px; text-align:center; margin:20px; color:#ff00aa; font-weight:bold;'>{self.current_kana}</div>"
        
        for i, btn in enumerate(self.buttons):
            btn.description = choices[i]
            btn.button_style = 'info'
            btn.disabled = False
        
        with self.feedback:
            clear_output()
        
        self.status.value = f"<b style='font-size:26px;color:#00ff00;'>Score: {self.score} &nbsp;&nbsp;🔥 Streak: {self.streak}</b>"
    
    def check_answer(self, b):
        if b.description == self.current_romaji:
            self.score += 10
            self.streak += 1
            b.button_style = 'success'
            with self.feedback:
                clear_output()
                print("🎉 YATTA! Correct! +10 points!")
                if self.streak % 5 == 0 and self.streak > 0:
                    print(f"🔥🔥🔥 {self.streak} STREAK! You're on fire! 🐼")
            self.fire_confetti()
        else:
            self.streak = 0
            b.button_style = 'danger'
            with self.feedback:
                clear_output()
                print(f"😅 It was **{self.current_romaji}**")
        
        for btn in self.buttons:
            btn.disabled = True
        self.status.value = f"<b style='font-size:26px;color:#00ff00;'>Score: {self.score} &nbsp;&nbsp;🔥 Streak: {self.streak}</b>"
    
    def speak(self, b):
        js = f"""
        const u = new SpeechSynthesisUtterance('{self.current_kana}');
        u.lang = 'ja-JP';
        speechSynthesis.speak(u);
        """
        display(Javascript(js))
    
    def fire_confetti(self):
        js = """
        (function() {
            const emojis = ['🎉','🐼','🌸','🍡','🔥','🦄','⭐','🍥'];
            for(let i = 0; i < 100; i++) {
                let c = document.createElement('div');
                c.style.position = 'fixed'; c.style.zIndex = 99999;
                c.style.left = Math.random()*100 + 'vw';
                c.style.top = '-30px';
                c.style.fontSize = (20 + Math.random()*30) + 'px';
                c.innerHTML = emojis[Math.floor(Math.random()*emojis.length)];
                document.body.appendChild(c);
                let dur = Math.random()*2000 + 2500;
                c.animate([{transform:'translateY(0) rotate(0deg)'}, 
                          {transform:`translateY(${window.innerHeight+100}px) rotate(${Math.random()*1200}deg)`}], 
                         {duration: dur, easing:'cubic-bezier(0.25,0.1,0.25,1)'});
                setTimeout(() => c.remove(), dur);
            }
        })();
        """
        display(Javascript(js))

# ==================== START THE GAME ====================
game = KanaBlast()
```
